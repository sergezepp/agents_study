```python
from agent import run_agent
from weatherTool import get_city_temperature

def get_weather_tool(tool_input):
    return get_city_temperature(tool_input["location"])

tools = [
    {
        "name": "get_weather",
        "description": "Get the current weather for a given location.",
        "input_schema": {
            "type": "object",
            "properties": {
                "location": {
                    "type": "string",
                    "description": "City and Country, e.g. Tokyo, Japan",
                }
            },
            "required": ["location"],
        },
    }
]
messages = [{"role": "user", "content": "What's the weather in Monterrey, Mexico?"}]

final_answer = run_agent(
    messages,
    tools=tools,
    tool_functions={"get_weather": get_weather_tool},
    system="You are a helpful weather assistant and you need to make recommendations about the weather",
)

print(final_answer)
```
