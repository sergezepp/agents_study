```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"

message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)

for block in message.content:
    if block.type == "text":
        print(block.text)
```

```python
from tools import add_assistant_message, add_user_message, chat

# Start with an empty message list
messages = []

# Add the initial user question
add_user_message(messages, "Define quantum computing in one sentence")

# Get Claude's response
answer = chat(messages)

# Add Claude's response to the conversation history
add_assistant_message(messages, answer)

# Add a follow-up question
add_user_message(messages, "Write another sentence")

# Get the follow-up response with full context
final_answer = chat(messages)

add_assistant_message(messages, final_answer)

for msg in messages:
    print(f"{msg['role'].capitalize()}: {msg['content']}")
```

```python
from dotenv import load_dotenv
load_dotenv()

from agent import run_agent
from weatherTool import get_city_temperature

def get_weather_tool(tool_input):
    return get_city_temperature(tool_input["location"])

tools = [
    {
        "name": "get_weather",
        "description": "Get the current weather for a given location.",
        "input_schema": {
            "type": "object",
            "properties": {
                "location": {
                    "type": "string",
                    "description": "City and Country, e.g. Tokyo, Japan",
                }
            },
            "required": ["location"],
        },
    }
]
messages = [{"role": "user", "content": "What's the weather in Reos, Vet?"}]

final_answer = run_agent(
    messages,
    tools=tools,
    tool_functions={"get_weather": get_weather_tool},
    system="You are a helpful weather assistant and you need to make recommendations about the weather",
)

print(final_answer)
```
